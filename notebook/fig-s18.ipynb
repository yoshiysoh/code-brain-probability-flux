# Hierarchical clustering by hypercubic probability traffic

Same clustering as `fig-03.ipynb`, applied to the probability traffic
T(mu, nu) = w(nu|mu) p(mu) + w(mu|nu) p(nu) instead of the probability flux.
Traffic is symmetric, so each of the 448 undirected hypercube edges
(7 * 2**7 / 2) is used once.

```python
import numpy as np

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT
panel_gap = bpp.PANEL_GAP_PT / 72  # gap between panels in the composite (in)
```

```python
pfa, cluster_inds = bpp.aggregated_pfa(data_dir=DATA_DIR)
num_clusters = len(np.unique(cluster_inds))
color = bpp.spin_colors(num_clusters)
pfa.perform_WPCA(spin_colors=color)
```

```python
traffic_matrices = bpp.traffic_matrices_by_task(pfa)
row_indices, col_indices = bpp.get_one_bit_difference_indices_undirected(
    num_clusters
)
sampled_traffic_flat = traffic_matrices[:, row_indices, col_indices]
```

```python
vmax = np.abs(sampled_traffic_flat).max()
bpp.plot_task_clustering(
    sampled_traffic_flat,
    "../figures/fig-s18",
    "Probability traffic",
    (max_fig_width - 3 * panel_gap) / 4,
    pfa.task_name_list,
    vmax=vmax,
    formats=("svg", "pdf", "png"),
)
```
