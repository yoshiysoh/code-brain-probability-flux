```python
import matplotlib.pyplot as plt

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT
```

```python
# Coarse-grain by the seven Yeo networks instead of the parcel clusters
pfa, network_ids = bpp.aggregated_pfa("yeo7", DATA_DIR)
_, network_names = bpp.yeo_networks()
color = bpp.spin_colors(len(network_names), plt.cm.Dark2)
pfa.perform_WPCA(spin_colors=color)
```

```python
fig, axs = plt.subplots(
    3, 3, figsize=(max_fig_width, max_fig_height), sharex=True, sharey=True
)
bpp.plot_pca_by_task(pfa, axs, kind="flux", scale=40, color=color)
bpp.save_panel(fig, "../figures/fig-s03", "fig-s03", bbox_inches=None)
```
