```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import make_axes_locatable

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT
```

```python
# Flux on a grid in the space of the first two principal components of the
# parcel activity (Lynn et al., 2021), with bootstrap confidence ellipses
# (group-level results of scripts/compute_aggregates.py)
pfa = bpp.PFA()
bin_edges, bin_pos, grid_summaries = bpp.load_pc_grid(DATA_DIR)
```

```python
fig, axs = plt.subplots(
    3, 3, figsize=(max_fig_width, max_fig_height), sharex=True, sharey=True
)
for k, ax in enumerate(bpp.task_axes(axs)):
    pfa.set_grid_summary(bin_edges, bin_pos, grid_summaries[k])
    ax, cbar = pfa.plot_flux_with_confidence(
        fig,
        ax,
        bootstrap=False,
        plot_cbar=False,
        vmax=0.048,
        label_pos=[1.02, 1.5, 1.4, 1.3],
    )
    ax.set_title(pfa.task_name_list[k])
    if k == 1:
        legend_ax = axs[0, 2]
        legend_ax.axis("off")
        cax = make_axes_locatable(legend_ax).append_axes(
            "left", size=0.08, pad=0.08
        )
        cb = fig.colorbar(cbar, cax=cax, label="Probability")
        cb.outline.set_visible(False)
fig.supxlabel("PC1")
fig.supylabel("PC2")
bpp.save_panel(fig, "../figures/fig-s02", "fig-s02", bbox_inches=None)
```
