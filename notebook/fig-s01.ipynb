```python
import matplotlib.pyplot as plt
import numpy as np

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT
```

```python
pfa, cluster_inds = bpp.aggregated_pfa(data_dir=DATA_DIR)
num_clusters = len(np.unique(cluster_inds))
color = bpp.spin_colors(num_clusters)
pfa.perform_WPCA(spin_colors=color)
```

```python
# Net probability change of each state, dp_mu/dt = sum_nu J(mu, nu), in
# units of its standard deviation over all tasks
flux_matrices = bpp.flux_matrices_by_task(pfa)
probability_change = flux_matrices.sum(axis=2)
probability_change = probability_change / np.std(probability_change)
```

```python
fig, axs = plt.subplots(
    3, 3, figsize=(max_fig_width, max_fig_width / 2), sharex=True, sharey=True
)
task_colors = bpp.task_colors_stepped()
bin_edges = np.linspace(-4, 4, 32)
for k, ax in enumerate(bpp.task_axes(axs)):
    ax.hist(probability_change[k], bins=bin_edges, color=task_colors[k])
    ax.set_title(pfa.task_name_list[k])
axs[0, 2].axis("off")
fig.supxlabel(
    r"Probability change with time $\Delta p_{\mu} / \sigma_{\Delta p_{\mu}}$"
)
fig.supylabel("Count")
bpp.save_panel(fig, "../figures/fig-s01", "fig-s01", bbox_inches=None)
```
