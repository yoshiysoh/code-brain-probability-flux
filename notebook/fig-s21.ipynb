```python
import matplotlib.pyplot as plt
import numpy as np

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT
```

```python
# This figure is dense enough that the project default (7pt) causes
# text/content overlap and excess empty space; use 5pt here instead.
plt.rcParams["font.size"] = 5.0
```

```python
# Panels a-f show one example recording; they need the HCP recordings and
# are made by `python scripts/hcp_example_panels.py`.
save_dir = "../figures/fig-s21"
settings = ["threshold_0.70", "threshold_0.70_order1", "threshold_0.70_order2"]
```

```python
# Panels g-i: binarization at the zeros of the z score's 0th, 1st and 2nd
# derivative
for order, setting in enumerate(settings):
    pfa, cluster_inds = bpp.aggregated_pfa(setting, DATA_DIR)
    color = bpp.spin_colors(len(np.unique(cluster_inds)))
    pfa.perform_WPCA(spin_colors=color)
    fig, axs = plt.subplots(
        3,
        3,
        figsize=(max_fig_width / 3, max_fig_height / 3 - 0.09),
        sharex=True,
        sharey=True,
    )
    bpp.plot_pca_by_task(pfa, axs, kind="traffic", scale=20, color=color)
    bpp.save_panel(fig, save_dir, "ghi"[order], bbox_inches="tight")
```
