```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

import brain_probability_flux.plotting as bpp
```

```python
# Results are read from data/. To plot a fresh run, set DATA_DIR to its run
# directory, for example Path("../data/2026-10-20").
DATA_DIR = bpp.DATA_DIR
```

```python
max_fig_width = bpp.FIG_WIDTH
max_fig_height = bpp.FIG_HEIGHT

TASK_NAMES = bpp.TASK_NAMES_SHORT
NUM_TASKS = 8
NUM_SPINS = 7
OUT_DIR = Path("../figures/fig-s12")
```

```python
# Arrhenius model with a spin-independent (SI, Fig. 4) and a spin-dependent
# (SD) rate constant, and their trajectory bootstrap
# (scripts/bootstrap_rate_constants.py)
A_sd = bpp.load_results("inferred_networks", "arrhenius_spindep", DATA_DIR)[
    "rate_const"
]  # (8, 7)
boot = bpp.load_results("rate_constant_bootstrap", "arrhenius", DATA_DIR)
A_sd_boot = boot["A_sd"]  # (8, 7, n_boot)
r_boot = {key: boot[f"r_{key}"] for key in ["SI", "SD"]}
loss_boot = {key: boot[f"loss_{key}"] for key in ["SI", "SD"]}
print("Bootstrap r  SI (median):", np.median(r_boot["SI"], axis=1).round(4))
print("Bootstrap r  SD (median):", np.median(r_boot["SD"], axis=1).round(4))
```

```python
# Bars of panel a: point estimates of A_k. Bootstrap medians are ~35% lower
# (Jensen-inequality bias from log-transforming Poisson counts), so the
# error bars place the bootstrap IQR half-widths around the point estimate.
fig, axs = plt.subplots(
    1, 3, figsize=(max_fig_width, max_fig_height / 3 + 0.2)
)
bpp.plot_rate_constant_comparison(
    axs,
    A_sd,
    A_sd_boot,
    loss_boot,
    r_boot,
    TASK_NAMES,
    A_label=r"Rate constant $A_k$",
    A_ylim=(0, 0.22),
    r_ylim=(0, 1.15),
)
bpp.save_axes_panels(fig, dict(zip("abc", axs)), OUT_DIR)
```
